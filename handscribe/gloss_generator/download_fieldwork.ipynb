```python
import os
import pandas as pd
from datasets import load_dataset
from collections import defaultdict
```

```python
def format_dataset(gt_sents, gt_translations, sep="___"):
    strings = [t + sep + g for t, g in zip(gt_sents, gt_translations)]
    return pd.DataFrame(strings, columns=[f"text{sep}gloss"])
```

```python
dataset = load_dataset("wav2gloss/fieldwork", cache_dir=OUT_FOLDER)
```

```python
train_data = dataset["train"]
train_lang, train_transcr, train_glosses = train_data['language'], train_data['transcription'], train_data['gloss']
print(f'We have {len(train_lang)} languages, {len(train_transcr)} transcriptions and {len(train_glosses)} glosses in the test set.')
```

```python
train_df = format_dataset(train_transcr, train_glosses)
print(f"We have {len(train_df)} rwos in the df.")
if not os.path.exists(os.path.join(OUT_FOLDER, "wav2gloss__train.csv")):
    train_df.to_csv(os.path.join(OUT_FOLDER, "wav2gloss__train.csv"), index=False)
```

```python
lang_to_datat = defaultdict(lambda: {"transcription": [], "gloss": []})

for lang, trans, gloss in zip(train_lang, train_transcr, train_glosses):
    lang_to_datat[lang]["transcription"].append(trans)
    lang_to_datat[lang]["gloss"].append(gloss)
    
print(f"We have {len(lang_to_datat)} languages in the dataset.")

for lang, data in lang_to_datat.items():
    num_examples = len(data["transcription"])
    avg_trans_len = sum(len(t.split()) for t in data["transcription"]) / num_examples
    avg_gloss_len = sum(len(g.split()) for g in data["gloss"]) / num_examples
    
    print(f"{lang} [{LANGS_DICT[lang]}] => {num_examples} Examples:  | Avg transcription len: {avg_trans_len:.2f} words | Avg gloss len: {avg_gloss_len:.2f} words")
```

```python
test_data = dataset["test"]
languages, transcriptions, glosses = test_data['language'], test_data['transcription'], test_data['gloss']
print(f'We have {len(languages)} languages, {len(transcriptions)} transcriptions and {len(glosses)} glosses in the test set.')
```

```python
test_df = format_dataset(transcriptions, glosses)
print(f"We have {len(test_df)} rwos in the df.")
if not os.path.exists(os.path.join(OUT_FOLDER, "wav2gloss__test.csv")):
    test_df.to_csv(os.path.join(OUT_FOLDER, "wav2gloss__test.csv"), index=False)
```

```python
lang_to_data = defaultdict(lambda: {"transcription": [], "gloss": []})

for lang, trans, gloss in zip(languages, transcriptions, glosses):
    lang_to_data[lang]["transcription"].append(trans)
    lang_to_data[lang]["gloss"].append(gloss)
    
print(f"We have {len(lang_to_data)} languages in the dataset.")

for lang, data in lang_to_data.items():
    num_examples = len(data["transcription"])
    avg_trans_len = sum(len(t.split()) for t in data["transcription"]) / num_examples
    avg_gloss_len = sum(len(g.split()) for g in data["gloss"]) / num_examples
    
    print(f"{lang} => {num_examples} Examples:  | Avg transcription len: {avg_trans_len:.2f} words | Avg gloss len: {avg_gloss_len:.2f} words")
```

```python
lang_to_df = {}

for lang, data in lang_to_data.items():
    df = format_dataset(data["transcription"], data["gloss"])
    lang_to_df[lang] = df

print(f"We have {len(lang_to_df)} dfs.")
```

```python
for key, value in lang_to_df.items():
    value.to_csv(os.path.join(OUT_CSV_DIR, key + ".csv"), index=False)
print(f"Stored {len(lang_to_df)} csv files in {OUT_CSV_DIR}.")
```

```python
SEEN_LANGS_CSV = os.path.join(OUT_CSV_DIR, "seen")
UNSEEN_LANGS_CSV = os.path.join(OUT_CSV_DIR, "unseen")
```

```python
for lang, data in lang_to_data.items():
    if lang in TRAIN_LANGS:
        print(f"Language {lang} found in train set with: {len(data['transcription'])} examples.")
        os.rename(os.path.join(OUT_CSV_DIR, lang + ".csv"), os.path.join(SEEN_LANGS_CSV, lang + ".csv"))
    elif lang in TEST_LANGS:
        print(f"Language {lang} found in test set with: {len(data['transcription'])} examples.")
        os.rename(os.path.join(OUT_CSV_DIR, lang + ".csv"), os.path.join(UNSEEN_LANGS_CSV, lang + ".csv"))
    else:
        print(f"Language {lang} not found in either train or test sets.")
        
print(f"Stored:\n-{len(os.listdir(SEEN_LANGS_CSV))} csvs in {SEEN_LANGS_CSV}.\n-{len(os.listdir(UNSEEN_LANGS_CSV))} csvs in {UNSEEN_LANGS_CSV}.")
```

```python
train_rows = []
test_rows = []

for lang, data in lang_to_data.items():
    if lang in TRAIN_LANGS:
        train_rows.extend(
            [t + "___" + g for t, g in zip(data["transcription"], data["gloss"])]
        )
    elif lang in TEST_LANGS:
        test_rows.extend(
            [t + "___" + g for t, g in zip(data["transcription"], data["gloss"])]
        )
    else:
        print(f"Language {lang} not found in either train or test sets.")

train_df = pd.DataFrame(train_rows, columns=["text___gloss"])
test_df = pd.DataFrame(test_rows, columns=["text___gloss"])
print(f"We have {len(train_df)} train rows and {len(test_df)} test rows.")
```

```python
train_df.to_csv(os.path.join(OUT_CSV_DIR, "wav2gloss__seen_lang.csv"), index=False)
test_df.to_csv(os.path.join(OUT_CSV_DIR, "wav2gloss__unseen_lang.csv"), index=False)
```
